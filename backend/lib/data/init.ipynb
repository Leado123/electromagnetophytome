```python
from datetime import datetime
from hrid import HRID

hruuid = HRID()
import h5py
# Establish hdf5 file hierarchy
file = h5py.File("../../../data/data.h5", "a")
```

```python
from datetime import datetime
import h5py
import numpy as np

file.require_dataset("sequencer", shape=(2,12,0), maxshape=(2,12,None), dtype=h5py.string_dtype(encoding='utf-8'))
file.require_dataset("sequencer_effects", shape=(2,12,0), maxshape=(2,12,None), dtype=h5py.string_dtype(encoding='utf-8'))

plant_group = file.require_group("plants")
analysis_group = file.require_group("analysis")
emp_group = file.require_group("emp")

def create_plant():
    p = plant_group.require_group(hruuid.generate())
    p.attrs["added_timestamp"] = datetime.now().isoformat()
    if "EMPResponse" not in p.keys():
        p.create_dataset("EMPResponse", shape=(0,), maxshape=(None,), dtype='f8')
    images = p.require_group("Images")
    if "images" not in images.keys():
        images.create_dataset("images", shape=(0, 1980, 1980, 3), maxshape=(None, 1980, 1980, 3), dtype='uint8')
    if "index" not in images.keys():
        images.create_dataset("index", shape=(0,), dtype='i8', maxshape=(None,))
    if "DepthScans" not in p.keys():
        p.create_dataset("DepthScans", shape=(0,), maxshape=(None,), dtype='f8')
    if "PlantGrowth" not in p.keys():
        p.create_dataset("PlantGrowth", shape=(0,), maxshape=(None,), dtype='f8')
    return p

def create_empID(plant_id: str):
    empID = emp_group.require_group(hruuid.generate())
    empID.attrs["plant_id"] = plant_id
    empID.attrs["timestamp"] = datetime.now().isoformat()
    parameters = empID.require_group("parameters")
    if "time" not in parameters.keys():
        parameters.create_dataset("time", shape=(0,), maxshape=(None,), dtype='f8')
    if "values" not in parameters.keys():
        parameters.create_dataset("values", shape=(0,), maxshape=(None,), dtype='f8')
    if "timestamps" not in empID.keys():
        empID.create_dataset("timestamps", shape=(0,), dtype=h5py.string_dtype(encoding='utf-8'), maxshape=(None,))
    plant_responses = empID.require_group("plant_responses")
    if "time" not in plant_responses.keys():
        plant_responses.create_dataset("time", shape=(0,), dtype='f8', maxshape=(None,))
    if "values" not in plant_responses.keys():
        plant_responses.create_dataset("values", shape=(0,), dtype='f8', maxshape=(None,))
    return empID

analysis_group.require_dataset("growth_points", shape=(0,), maxshape=(None,), dtype="f")
```

```python
from datetime import datetime
import h5py

try:
    _ = file.filename
except:
    file = h5py.File("../../../data/data.h5", "a")
    plant_group = file.require_group("plants")
    emp_group = file.require_group("emp")

create_plant()
create_empID("plant_id")

file.close()
```
